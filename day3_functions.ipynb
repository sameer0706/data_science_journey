```python
def check_result(m):
    if(m>=40):
        return "pass"
    else:
        return "fail"
print(check_result(39))
print(check_result(80))
```

```python
def negative(nums):
    return [x for x in nums if x<0]
numbers=[-2,5,-9,-11,4,3]
print(negative(numbers))
#list comprehension -> extracting a list from existing list
```

```python
def negativesquares(nums):
    return [x*x for x in nums if x<0]
numbers=[-2,5,-9,-11,4,3]
print(negativesquares(numbers))
```

```python
#default argument[pass mark]
def passed(m,p=50):
    if(m>=p):
        return "pass"
    else:
        return "fail"
print(passed(60))
print(passed(48,40))
print(passed(59,60))
```

```python
def passed(m,p=50):
    if(m>=p):
        return 1
    else:
        return 0
marks=[35,88,38,44,56,67,77]
r50=[passed(n) for n in marks ]
pc=sum(r50)
fc=len(r50)-pc
print(r50, pc, fc)
print([passed(n,40) for n in marks])
```
