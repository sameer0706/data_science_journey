```python
def grade(n):
    if n>=90:
        return "O"
    elif n>=80:
        return "A"
    elif n>=70:
        return "B"
    elif n>=60:
        return "C"
    elif n>=50:
        return "D"
    else:
        return "F"
print(grade(75))
print(grade(39))
```
