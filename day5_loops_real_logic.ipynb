```python
def countingpn(n):
    pc=0
    nc=0
    for x in n :
        if x>0:
            pc=pc+1
        elif x<0:
            nc=nc+1
    return pc,nc
nums=[1,-1,2,3,-2,-3,-4,-5]
p,n=countingpn(nums)
print(p,n,len(nums)-n-p,)
#counting postives and negatives
```

```python
def mini(num):
    mn=num[0]
    mx=num[0]
    for x in num:
        if mn>x:
            mn=x
        if mx<x:
            mx=x
    return mn,mx
arr=[-1,-2,3,4,56,0,-45]
x,y=mini(arr)
print(x,y)
#min and max
```

```python
def valid(n):
    r=[]
    for x in n:
        if x>=0 and x<=100:
            r.append(x)
    return r
marks = [45, 67, -10, 89, 105, 72]
print(valid(marks))
#validating marks
```

```python
def valid(n):
    result=[x for x in n if x>=0 and x<=100]
    return result
marks = [45, 67, -10, 89, 105, 72]
print(valid(marks))
```

```python
#second largest
def secondmax(num):
    fmx=float('-inf')
    smx=float('-inf')
    for x in num:
        if x>fmx:
            smx=fmx
            fmx=x
        elif x>smx:
            smx=x
    return fmx,smx
arr=[-1,-2,3,4,56,0,10,-45]
x,y=secondmax(arr)
print(x,y)
```
