```python
fruits=["apple","banana","apple","apple", "orange","banana"]
fruits_count ={}
for x in fruits:
    if fruits_count.get(x,"n")=="n":
        fruits_count[x]=1
    else:
        fruits_count[x]=fruits_count[x]+1
print(fruits_count)
```

```python
def freq_counting(d):
    fc={}
    for x in d:
        fc[x]=fc.get(x,0)+1
    return fc
ff=["apple","banana","apple","apple", "orange","banana"]
print(freq_counting(ff))
```
